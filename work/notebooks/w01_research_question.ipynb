## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
from pathlib import Path
import pandas as pd

root = Path.cwd()
candidate_paths = [root, root.parent, root.parent.parent]
csv_path = None
for candidate in candidate_paths:
    potential_path = candidate / 'data' / 'raw' / 'content_refresh_anonymized.csv'
    if potential_path.exists():
        csv_path = potential_path
        break

if csv_path is None:
    raise FileNotFoundError('Could not find the starter CSV.')

df = pd.read_csv(csv_path)

overall_down_rate = (df['trend_direction'] == 'down').mean()
high_impression_down_rate = df.loc[df['impressions_90d'] > 1000, 'trend_direction'].eq('down').mean()
visible_decline_rate = df.loc[(df['impressions_90d'] > 1000) & (df['avg_position'] > 0) & (df['avg_position'] <= 10), 'trend_direction'].eq('down').mean()
low_ctr_decline_rate = df.loc[(df['impressions_90d'] > 2000) & (df['ctr'] < 0.5), 'trend_direction'].eq('down').mean()

print('Overall decline share:', round(overall_down_rate * 100, 1), '%')
print('Pages with >1,000 impressions and a downward trend:', round(high_impression_down_rate * 100, 1), '%')
print('Pages with >1,000 impressions and page-1 position:', round(visible_decline_rate * 100, 1), '%')
print('Pages with >2,000 impressions and low CTR:', round(low_ctr_decline_rate * 100, 1), '%')

print('These slices suggest that visible pages with weak recent signals are not random; they show a meaningful pattern worth studying.')
```
