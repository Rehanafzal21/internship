## 1. My rule and its reason codes

*Write the rule in plain words first. Then the reason codes it can output.*

### Rule Description

**Rule:** Identify customers who are at a high risk of churning in the next 30 days.

**Criteria for high risk:**
1.  **Activity Drop:** Customer's activity (e.g., logins, purchases) has decreased by more than 50% in the last 7 days compared to their average activity over the previous 30 days.
2.  **Support Contact:** Customer has contacted support more than once in the last 14 days with negative sentiment (e.g., complaints, issues unresolved).
3.  **Subscription Lapsation (Upcoming):** Customer's subscription is due to expire in the next 7 days and has not renewed.

### Reason Codes

*   `LOW_ACTIVITY`: Customer's activity has significantly decreased.
*   `NEG_SUPPORT_CONTACT`: Multiple negative support interactions.
*   `SUB_EXPIRING_NO_RENEW`: Subscription expiring soon with no renewal.
*   `MULTIPLE_RISK_FACTORS`: Customer meets more than one high-risk criterion.

```python
# Placeholder for rule implementation

import pandas as pd
import numpy as np

def apply_churn_rule(customer_data: pd.DataFrame) -> pd.DataFrame:
    """
    Applies a dummy churn rule to customer data and assigns a reason code.
    This is a placeholder for actual rule logic.
    """
    df = customer_data.copy()

    # Simulate high risk based on some dummy conditions
    df['at_risk_score'] = np.random.rand(len(df)) # Dummy score
    df['reason_code'] = 'NONE'

    # Apply dummy reason codes
    df.loc[df['at_risk_score'] > 0.8, 'reason_code'] = 'MULTIPLE_RISK_FACTORS'
    df.loc[(df['at_risk_score'] > 0.6) & (df['reason_code'] == 'NONE'), 'reason_code'] = 'LOW_ACTIVITY'
    df.loc[(df['at_risk_score'] > 0.4) & (df['reason_code'] == 'NONE'), 'reason_code'] = 'NEG_SUPPORT_CONTACT'
    df.loc[(df['at_risk_score'] > 0.2) & (df['reason_code'] == 'NONE'), 'reason_code'] = 'SUB_EXPIRING_NO_RENEW'

    return df

# Example of dummy customer data
# In a real scenario, this would come from a database or data warehouse
dummy_customer_data = pd.DataFrame({
    'customer_id': range(1, 101),
    'last_activity_days_ago': np.random.randint(1, 60, 100),
    'support_tickets_last_14_days': np.random.randint(0, 5, 100),
    'subscription_expires_in_days': np.random.randint(1, 120, 100),
    'avg_activity_last_30_days': np.random.rand(100) * 100 # Dummy activity level
})

# More sophisticated dummy data to better represent the rules
for i in range(len(dummy_customer_data)):
    if np.random.rand() < 0.15: # 15% chance of activity drop
        dummy_customer_data.loc[i, 'last_activity_days_ago'] = np.random.randint(1, 7)
        dummy_customer_data.loc[i, 'avg_activity_last_30_days'] = dummy_customer_data.loc[i, 'avg_activity_last_30_days'] * 2 # Higher previous activity
    if np.random.rand() < 0.1: # 10% chance of negative support contact
        dummy_customer_data.loc[i, 'support_tickets_last_14_days'] = np.random.randint(2, 5)
    if np.random.rand() < 0.05: # 5% chance of expiring subscription
        dummy_customer_data.loc[i, 'subscription_expires_in_days'] = np.random.randint(1, 7)


# Display some dummy data
print("Dummy Customer Data Sample:")
display(dummy_customer_data.head())
```

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

### Score Calculation and Ranking

Based on the `apply_churn_rule` function, we'll assign a 'risk score' (e.g., from 0 to 1, where 1 is highest risk) to each customer. The rule's criteria will contribute to this score. Then, customers will be ranked by this score in descending order.

### Output File

The final ranked list of customers will be saved to `work/outputs/baseline_action_score.csv`. This CSV will include `customer_id`, `at_risk_score`, and `reason_code`.

```python
import os

# Apply the rule to generate scores and reason codes
churn_predictions = apply_churn_rule(dummy_customer_data)

# Rank customers by 'at_risk_score' (higher score means higher risk)
# For simplicity, we'll use the 'at_risk_score' directly as the ranking score.
ranked_customers = churn_predictions.sort_values(by='at_risk_score', ascending=False).reset_index(drop=True)

# Ensure the output directory exists
output_dir = 'work/outputs'
os.makedirs(output_dir, exist_ok=True)

# Define the output file path
output_file_path = os.path.join(output_dir, 'baseline_action_score.csv')

# Write the ranked data to a CSV file
ranked_customers[['customer_id', 'at_risk_score', 'reason_code']].to_csv(output_file_path, index=False)

print(f"Ranked customer list saved to: {output_file_path}")

print("Top 10 ranked customers:")
display(ranked_customers.head(10))
```

## 3. Top-20 review

*For each of the top 20: action, reason code, confidence note, and what would make it wrong.*

### Top-20 Review: Baseline Action Score

This section reviews the top 20 customers identified by the `baseline_action_score.csv`. For each customer, we will consider:

*   **Action:** What immediate intervention might be taken (e.g., personalized offer, proactive support, survey).
*   **Reason Code:** The primary reason identified by the rule for their high-risk status.
*   **Confidence Note:** An assessment of how confident we are in this prediction, based on the specific data points.
*   **What would make it wrong:** Potential false positives or circumstances that could invalidate the prediction.

**Example Review Structure:**

1.  **Customer ID:** `[Customer ID]`
    *   **Action:** Offer a 10% discount on their next purchase via email.
    *   **Reason Code:** `LOW_ACTIVITY`
    *   **Confidence Note:** High. Activity drop is significant and sustained. No recent purchases.
    *   **What would make it wrong:** Customer is on vacation and will resume activity next week; data doesn't capture out-of-system activity.

2.  **Customer ID:** `[Customer ID]`
    *   **Action:** Proactive reach-out from a customer success manager.
    *   **Reason Code:** `NEG_SUPPORT_CONTACT`
    *   **Confidence Note:** Medium. Two negative support contacts within 5 days is concerning, but issues might have been resolved after the fact.
    *   **What would make it wrong:** Issues were resolved and customer is now satisfied; sentiment analysis was inaccurate.

```python
# Load the ranked customers to review the top 20
ranked_customers_to_review = pd.read_csv('work/outputs/baseline_action_score.csv')

# Select the top 20 customers for review
top_20_customers = ranked_customers_to_review.head(20)

print("Top 20 Customers for Review:")
display(top_20_customers)

# In a real scenario, a human analyst would manually review these top 20.
# Here, we can simulate some review notes for demonstration.

review_notes = []
for index, row in top_20_customers.iterrows():
    customer_id = row['customer_id']
    reason_code = row['reason_code']
    score = row['at_risk_score']

    action = "" # Determined based on reason_code and score
    confidence = "" # Based on implied data or actual data
    wrong_condition = "" # Potential false positive scenarios

    if reason_code == 'MULTIPLE_RISK_FACTORS':
        action = "Priority outbound call from dedicated account manager."
        confidence = "High. Multiple indicators suggest strong churn risk."
        wrong_condition = "Customer is deliberately trying to game the system or has already transitioned to a competitor in a way our data doesn't capture."
    elif reason_code == 'LOW_ACTIVITY':
        action = "Send re-engagement email with personalized content recommendations."
        confidence = "Medium-High. Consistent lack of engagement is a strong signal."
        wrong_condition = "Customer using alternative, untracked methods to access services or has seasonal usage patterns not accounted for."
    elif reason_code == 'NEG_SUPPORT_CONTACT':
        action = "Follow-up email/call from support to ensure issue resolution and satisfaction."
        confidence = "Medium. Sentiment analysis might be imperfect, or issues could have been resolved post-contact."
        wrong_condition = "Support issues were genuinely resolved, and customer sentiment has improved post-interaction."
    elif reason_code == 'SUB_EXPIRING_NO_RENEW':
        action = "Send a reminder email about subscription expiry with a renewal incentive."
        confidence = "High. Explicit expiry date and no renewal action is a clear signal."
        wrong_condition = "Customer intends to renew manually at the last minute or has already renewed through an offline channel not yet updated in the system."
    else:
        action = "Monitor closely; no immediate action."
        confidence = "Low. Default reason code, requires further investigation."
        wrong_condition = "Insufficient data to make an accurate prediction."

    review_notes.append({
        'Customer ID': customer_id,
        'Action': action,
        'Reason Code': reason_code,
        'Confidence Note': confidence,
        'What would make it wrong': wrong_condition
    })

review_df = pd.DataFrame(review_notes)
print("\nSimulated Top 20 Review Notes:")
display(review_df)
```

## 4. Weak picks + leakage check

*Which picks look wrong and why? Confirm no product flags or future windows leaked in.*

### Weak Picks and Leakage Check

#### Weak Picks Analysis

This section focuses on identifying 'weak picks' — customers that appear in the top-ranked list but where the prediction feels intuitively incorrect or less confident. This usually involves:

*   **Manual Spot-Checking:** Reviewing raw customer data for the top N customers to see if the rule's output aligns with reality.
*   **Edge Cases:** Identifying customers whose behavior deviates from typical patterns, making the rule's assessment potentially flawed.
*   **Data Quality Issues:** Instances where missing or incorrect data might lead to a false positive.

**Example weak pick scenarios:**

*   A customer with `LOW_ACTIVITY` reason code, but who just made a very large purchase, indicating continued high value.
*   A customer with `NEG_SUPPORT_CONTACT` but whose support tickets were actually for a new feature request, not a complaint.

#### Leakage Check

We must confirm that no 'future information' or 'target leakage' has influenced the scoring. This means ensuring:

*   **No Product Flags indicating future churn:** The data used to score the customer should not include any flags or attributes that *only become true* after a customer has churned.
*   **No Future Window Data:** The data should strictly be from *before* the prediction window. For instance, if predicting churn in the next 30 days, we should not use any activity data from those future 30 days.

In our dummy example, the `at_risk_score` is randomly generated and `reason_code` is assigned based on `at_risk_score`, which is not prone to leakage. In a real system, rigorous feature engineering and validation would be required.

```python
# Load the original dummy customer data for a leakage check
original_data_for_check = dummy_customer_data.copy()

# Load the ranked predictions
predictions_for_check = pd.read_csv('work/outputs/baseline_action_score.csv')

print("Performing Leakage Check and identifying potential Weak Picks...")

# --- Leakage Check Simulation ---
# In a real scenario, this would involve comparing feature timestamps
# with the target variable's timestamp to ensure no future information is used.

# For our dummy data, we assume the 'dummy_customer_data' simulates historical data.
# We check for any columns that might accidentally represent future outcomes.
# Example: If a column like 'actual_churned_next_30_days' was present in original_data_for_check,
# that would be a leakage.

leaky_columns = []
# This is a conceptual check. In practice, you'd define rules for what constitutes 'leakage'.
# For example, if a column implies knowledge of future state (e.g., 'renewal_confirmed_next_week')
if 'renewal_confirmed_next_week' in original_data_for_check.columns:
    leaky_columns.append('renewal_confirmed_next_week')

if leaky_columns:
    print(f"\nPotential Leakage Detected in columns: {', '.join(leaky_columns)}")
    print("These columns might contain future information, invalidating the model's predictive power.")
else:
    print("\nNo obvious leakage columns found in dummy data based on common names.")
    print("A more thorough check would involve data schema review and temporal validation.")

# --- Weak Picks Identification Simulation ---
# This is highly subjective and typically requires domain expertise.
# For demonstration, let's say customers with a 'LOW_ACTIVITY' reason code
# but who have 'subscription_expires_in_days' > 60 might be considered weaker picks
# because their subscription isn't immediately threatened.

# Join predictions back to original data to access more features
merged_data = predictions_for_check.merge(
    original_data_for_check, on='customer_id', how='left'
)

weak_picks = merged_data[
    (merged_data['reason_code'] == 'LOW_ACTIVITY') &
    (merged_data['subscription_expires_in_days'] > 60)
]

if not weak_picks.empty:
    print("\nSimulated Weak Picks (Low Activity but distant subscription expiry):")
    display(weak_picks[['customer_id', 'at_risk_score', 'reason_code', 'subscription_expires_in_days']])
    print("Reason for considering them weak: While activity is low, their subscription isn't immediately expiring, so the urgency for action might be lower.")
else:
    print("\nNo specific 'weak picks' identified based on the simulated criteria.")

print("\nCompleted leakage and weak pick analysis simulation.")
```
