## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

### Data Limitations

This dataset, while informative, has several limitations that restrict the types of questions it can answer fully:

1.  **Incomplete Data History (Unbalanced History):** The high percentage of missing values in `gsc_data_start` (36%) and `ga4_data_start` (53%) means that for many clients, we do not know when their Search Console or Analytics 4 data became available. This prevents a complete understanding of data coverage over time for all clients.

2.  **Varied Data Availability for GSC/GA4:** The presence of `access_profile` values like 'gsc_only', 'ga4_only', and 'no_search_or_analytics_access', combined with the high missingness in `gsc_data_start` and `ga4_data_start`, indicates that not all clients have both GSC and GA4 data. This limits analyses requiring integrated data from both sources for all clients.

3.  **Missing Core Status Information:** The 9% missing values in `is_active`, `has_gsc_access`, `has_ga4_access`, and `client_created_date` means that for a subset of clients, we lack fundamental information about their status or when their records began. This can introduce bias if these missing values are not random.

4.  **Temporal Overlap Ambiguity:** While `client_created_date` and `client_updated_date` cover a broad period, the start dates for `gsc_data_start` and `ga4_data_start` vary significantly (e.g., GSC data starts as early as January 2025, while GA4 data starts no earlier than October 2025). This implies that concurrent data from both sources is not guaranteed for all clients or for the entire observed period, complicating longitudinal studies or comparisons between GSC and GA4 metrics.

5.  **"Source Only Missing Client Dimension" Clients:** The `access_profile` category 'source_only_missing_client_dimension' (9% of clients) suggests a class of clients for whom detailed access information might be incomplete or undefined in the data source itself. This group might need to be excluded from certain analyses or handled with specific imputation strategies.

6.  **No Explicit Labels:** As noted in the field categorization, there isn't an explicit label (target variable) in this dataset. This means the data cannot directly answer predictive questions like "which clients are likely to churn" or "which clients will increase their data access," without defining and deriving such labels from existing features.
